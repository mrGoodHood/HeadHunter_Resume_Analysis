В результате преобразования у вас должен получиться столбец, содержащий информацию о том, сколько месяцев проработал соискатель.

```python
for value in hh_df['Опыт работы'].head(5):
    print(repr(value[:200]) if isinstance(value, str) else value)
    print()
```

```python
def get_experience(value):
    if pd.isna(value) or value.strip() == 'Не указано':
        return np.nan

    words = value.split()[2:]
    total_months = 0

    for i in (0, 2):
        if i + 1 >= len(words):
            break

        number, unit = words[i:i + 2]

        if not number.isdigit():
            break

        if unit in ('год', 'года', 'лет'):
            total_months += int(number) * 12
        elif unit in ('месяц', 'месяца', 'месяцев'):
            total_months += int(number)
        else:
            break

    return total_months


hh_df['Опыт работы (месяц)'] = hh_df['Опыт работы'].apply(get_experience)

hh_df['Опыт работы (месяц)'].head()
```

```python
million_cities = [
    'Новосибирск', 'Екатеринбург', 'Нижний Новгород',
    'Казань', 'Челябинск', 'Омск', 'Самара',
    'Ростов-на-Дону', 'Уфа', 'Красноярск',
    'Пермь', 'Воронеж', 'Волгоград'
]


def get_city(value):
    city = value.split(',')[0].strip()

    if city in ('Москва', 'Санкт-Петербург'):
        return city
    elif city in million_cities:
        return 'город-миллионник'
    else:
        return 'другие'


hh_df['Город'] = hh_df['Город, переезд, командировки'].apply(get_city)

hh_df['Город'].value_counts(dropna=False)
```

```python
def get_relocation(value):
    text = value.lower()

    if 'не готов к переезду' in text or 'не готова к переезду' in text:
        return False

    return (
        'готов к переезду' in text
        or 'готова к переезду' in text
        or 'хочу переехать' in text
    )


hh_df['Готовность к переезду'] = (
    hh_df['Город, переезд, командировки'].apply(get_relocation)
)

hh_df['Готовность к переезду'].value_counts(dropna=False)
```

```python
def get_business_trips(value):
    text = value.lower()

    if 'не готов к командировкам' in text or 'не готова к командировкам' in text:
        return False

    return (
        'готов к командировкам' in text
        or 'готова к командировкам' in text
        or 'готов к редким командировкам' in text
        or 'готова к редким командировкам' in text
    )


hh_df['Готовность к командировкам'] = (
    hh_df['Город, переезд, командировки'].apply(get_business_trips)
)

hh_df['Готовность к командировкам'].value_counts(dropna=False)
```

```python
spb_percent = (hh_df['Город'] == 'Санкт-Петербург').mean() * 100

both_percent = (
    hh_df['Готовность к переезду']
    & hh_df['Готовность к командировкам']
).mean() * 100

print(f'Живут в Санкт-Петербурге: {spb_percent:.2f}%')
print(f'Готовы к переезду и командировкам: {both_percent:.2f}%')
```

```python
employment_types = [
    'полная занятость', 'частичная занятость',
    'проектная работа', 'волонтерство', 'стажировка'
]

schedule_types = [
    'полный день', 'сменный график', 'гибкий график',
    'удаленная работа', 'вахтовый метод'
]

for category in employment_types:
    hh_df[category] = hh_df['Занятость'].str.contains(
        category, regex=False, na=False
    )

for category in schedule_types:
    hh_df[category] = hh_df['График'].str.contains(
        category, regex=False, na=False
    )

hh_df[employment_types + schedule_types].sum()
```

```python
rates_df = pd.read_csv('data/ExchangeRates.csv')

rates_df = rates_df[['currency', 'date', 'close', 'proportion']]

rates_df.head()
```

```python
rates_df['date'] = pd.to_datetime(
    rates_df['date'],
    format='%d/%m/%y'
)

hh_df['Обновление резюме'] = pd.to_datetime(
    hh_df['Обновление резюме'],
    format='%d.%m.%Y %H:%M'
).dt.normalize()

print(rates_df['date'].head())
print(hh_df['Обновление резюме'].head())
```

```python
salary_parts = hh_df['ЗП'].str.rsplit(n=1, expand=True)

hh_df['Сумма ЗП'] = pd.to_numeric(
    salary_parts[0].str.replace(' ', '', regex=False)
)
hh_df['Валюта ЗП'] = salary_parts[1]

print(hh_df[['ЗП', 'Сумма ЗП', 'Валюта ЗП']].head())
print('\nВалюты:')
print(hh_df['Валюта ЗП'].value_counts(dropna=False))
```

```python
currency_map = {
    'руб.': 'RUB',
    'KZT': 'KZT',
    'USD': 'USD',
    'бел.руб.': 'BYN',
    'EUR': 'EUR',
    'грн.': 'UAH',
    'сум': 'UZS',
    'KGS': 'KGS',
    'AZN': 'AZN'
}

hh_df['Валюта ЗП'] = hh_df['Валюта ЗП'].map(currency_map)

print(hh_df['Валюта ЗП'].value_counts(dropna=False))
print('\nКоды в таблице курсов:')
print(rates_df['currency'].unique())
```

```python
missing_values = hh_df.isna().sum()
print(missing_values[missing_values > 0])
```

3. Итак, у нас есть пропуски в 3ех столбцах: **"Опыт работы (месяц)"**, **"Последнее/нынешнее место работы"**, **"Последняя/нынешняя должность"**. Поступим следующим образом: удалите строки, где есть пропуск в столбцах с местом работы и должностью. Пропуски в столбце с опытом работы заполните **медианным** значением.

```python
hh_df = hh_df.dropna(
    subset=['Последнее/нынешнее место работы', 'Последняя/нынешняя должность']
)

experience_median = hh_df['Опыт работы (месяц)'].median()

hh_df['Опыт работы (месяц)'] = (
    hh_df['Опыт работы (месяц)'].fillna(experience_median)
)

print('Медиана для заполнения:', experience_median)
print('Осталось строк:', len(hh_df))
print('Всего пропусков:', hh_df.isna().sum().sum())
print('Средний опыт:', hh_df['Опыт работы (месяц)'].mean())
```

4. Мы добрались до ликвидации выбросов. Сначала очистим данные вручную. Удалите резюме, в которых указана заработная плата либо выше 1 млн. рублей, либо ниже 1 тыс. рублей.

```python
salary_outliers = (
    (hh_df['ЗП (руб)'] > 1_000_000)
    | (hh_df['ЗП (руб)'] < 1_000)
)

print('Строк для удаления:', salary_outliers.sum())

hh_df = hh_df.loc[~salary_outliers].copy()

print('Осталось строк:', len(hh_df))
```

```python
experience_outliers = (
    hh_df['Опыт работы (месяц)'] > hh_df['Возраст'] * 12
)

print('Строк для удаления:', experience_outliers.sum())
print(
    hh_df.loc[
        experience_outliers,
        ['Возраст', 'Опыт работы (месяц)']
    ]
)

hh_df = hh_df.loc[~experience_outliers].copy()

print('Осталось строк:', len(hh_df))
```

6. В результате анализа мы обнаружили потенциальные выбросы в признаке **"Возраст"**. Это оказались резюме людей чересчур преклонного возраста для поиска работы. Попробуйте построить распределение признака в **логарифмическом масштабе**. Добавьте к графику линии, отображающие **среднее и границы интервала метода трех сигм**. Напомним, сделать это можно с помощью метода axvline. Например, для построение линии среднего будет иметь вид:

`histplot.axvline(log_age.mean(), color='k', lw=2)`

В какую сторону асимметрично логарифмическое распределение? Напишите об этом в комментарии к графику.
Найдите выбросы с помощью метода z-отклонения и удалите их из данных, используйте логарифмический масштаб. Давайте сделаем послабление на **1 сигму** (возьмите 4 сигмы) в **правую сторону**.

Выведите таблицу с полученными выбросами и оцените, с каким возрастом соискатели попадают под категорию выбросов?

```python
log_age = np.log(hh_df['Возраст'])

mean_log = log_age.mean()
std_log = log_age.std()

lower_bound = mean_log - 3 * std_log
upper_bound = mean_log + 4 * std_log

fig, ax = plt.subplots(figsize=(10, 5))

sns.histplot(log_age, bins=30, color='mediumpurple', ax=ax)

ax.axvline(mean_log, color='black', label='Среднее')
ax.axvline(
    lower_bound, color='purple', linestyle='--',
    label='Нижняя граница: −3σ'
)
ax.axvline(
    upper_bound, color='purple', linestyle=':',
    label='Верхняя граница: +4σ'
)

ax.set_title('Распределение логарифма возраста')
ax.set_xlabel('Натуральный логарифм возраста')
ax.set_ylabel('Количество резюме')
ax.legend()

plt.tight_layout()
plt.show()

print('Асимметрия:', log_age.skew())
print('Нижняя граница возраста:', np.exp(lower_bound))
print('Верхняя граница возраста:', np.exp(upper_bound))
```

```python
age_outliers = (
    (log_age < lower_bound)
    | (log_age > upper_bound)
)

print('Выбросы:')
print(
    hh_df.loc[age_outliers, ['Возраст', 'Опыт работы (месяц)']]
    .sort_values('Возраст')
)

print('Строк для удаления:', age_outliers.sum())

hh_df = hh_df.loc[~age_outliers].copy()

print('Осталось строк:', len(hh_df))
```

Логарифмическое распределение возраста имеет правостороннюю асимметрию: коэффициент асимметрии составляет около 0,45. 
Для поиска выбросов использованы границы −3σ и +4σ относительно среднего логарифма возраста.

В исходной шкале границы составляют примерно 15,66 и 78,88 года. 
Следовательно, под критерий выбросов попадают резюме соискателей в возрасте 15 лет и младше, а также 79 лет и старше.
